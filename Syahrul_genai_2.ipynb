
# Assignment 2: Exploring Text Generation Using GPT-2

## Objective
The objective of this assignment is to understand how different text-generation parameters influence the output produced by a language model.

This notebook uses the pretrained **GPT-2** model from the Hugging Face Transformers library. The same opening sentence is used in five experiments while changing one generation parameter at a time.

The experiments cover:
- Temperature
- Top-k
- Top-p
- Repetition penalty

A custom prompt is also used to generate three different versions.


## 1. Install and Import Required Libraries

```python
!pip -q install transformers torch
```

```python

import torch
from transformers import pipeline, set_seed

generator = pipeline("text-generation", model="gpt2")
print("GPT-2 model loaded successfully.")
```


## 2. Original Prompt

The same opening sentence is used for the five experiments so that the effect of the changed generation parameter can be compared.


```python

original_prompt = "The old bookstore at the end of the street had a secret that nobody knew about."
print("Original prompt:")
print(original_prompt)
```

## 3. Experiment 1 - Temperature

**Parameter changed:** Temperature  
**Value:** 0.7

```python

set_seed(42)
output = generator(original_prompt, max_new_tokens=80, do_sample=True,
                   temperature=0.7, pad_token_id=generator.tokenizer.eos_token_id)[0]["generated_text"]
print(output)
```

**Observation:** A moderate temperature allows some variation while generally keeping the output coherent.

## 4. Experiment 2 - Top-k

**Parameter changed:** Top-k  
**Value:** 20

```python

set_seed(42)
output = generator(original_prompt, max_new_tokens=80, do_sample=True,
                   top_k=20, pad_token_id=generator.tokenizer.eos_token_id)[0]["generated_text"]
print(output)
```

**Observation:** Restricting choices to the 20 most likely tokens can make the text more focused and predictable.

## 5. Experiment 3 - Top-p

**Parameter changed:** Top-p  
**Value:** 0.85

```python

set_seed(42)
output = generator(original_prompt, max_new_tokens=80, do_sample=True,
                   top_p=0.85, pad_token_id=generator.tokenizer.eos_token_id)[0]["generated_text"]
print(output)
```

**Observation:** Top-p provides a probability-based set of candidate tokens and can balance creativity with coherence.

## 6. Experiment 4 - Repetition Penalty

**Parameter changed:** Repetition penalty  
**Value:** 1.5

```python

set_seed(42)
output = generator(original_prompt, max_new_tokens=80, do_sample=True,
                   repetition_penalty=1.5, pad_token_id=generator.tokenizer.eos_token_id)[0]["generated_text"]
print(output)
```

**Observation:** A higher repetition penalty discourages the model from repeatedly selecting the same tokens or phrases.

## 7. Experiment 5 - Lower Temperature

**Parameter changed:** Temperature  
**Value:** 0.3

```python

set_seed(42)
output = generator(original_prompt, max_new_tokens=80, do_sample=True,
                   temperature=0.3, pad_token_id=generator.tokenizer.eos_token_id)[0]["generated_text"]
print(output)
```

**Observation:** A lower temperature makes the generation more conservative and predictable.


## 8. Comparison of the Five Experiments

| Experiment | Parameter Changed | Value | Main Effect |
|---|---|---:|---|
| 1 | Temperature | 0.7 | Moderate creativity and coherence |
| 2 | Top-k | 20 | More focused token choices |
| 3 | Top-p | 0.85 | Balance between variation and coherence |
| 4 | Repetition penalty | 1.5 | Reduced repetition |
| 5 | Temperature | 0.3 | More predictable and conservative text |

### Overall Analysis
The generation parameters affect how GPT-2 selects the next token. Lower temperature generally produces more predictable text, while higher temperature allows more variation. Top-k restricts choices to a fixed number of likely tokens, while top-p uses a probability-based group. Repetition penalty discourages repeated tokens.

For predictable text, lower temperature is useful. For more creative writing, a moderate or higher temperature can provide more variation, although excessive randomness can reduce coherence.

## 11. Conclusion

This experiment demonstrated how GPT-2 generation parameters affect generated text. Temperature controls randomness, top-k limits the number of likely token choices, top-p controls the cumulative probability of candidate tokens, and repetition penalty discourages repeated tokens.

Overall, more restrictive settings tend to produce predictable text, while settings that allow more variation can produce more creative outputs. Excessive randomness, however, can reduce coherence.
